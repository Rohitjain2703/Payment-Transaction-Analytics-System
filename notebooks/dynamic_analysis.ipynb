```python

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Find project folder automatically
current_folder = Path.cwd()

project_root = None

for folder in [current_folder, *current_folder.parents]:
    if (folder / "data" / "raw").exists():
        project_root = folder
        break

if project_root is None:
    raise FileNotFoundError(
        "Project folder not found. "
        "Please open the notebook inside the project."
    )

# Load datasets
data_folder = project_root / "data" / "raw"

payment_methods = pd.read_csv(
    data_folder / "payment_methods.csv"
)

transactions = pd.read_csv(
    data_folder / "transactions.csv"
)

print("Project folder:", project_root)
print("Payment methods:", payment_methods.shape)
print("Transactions:", transactions.shape)

print("Datasets loaded successfully!")
```

```python

def check_data_quality(data, dataset_name):
    print(f"\n{'=' * 40}")
    print(f"DATA QUALITY REPORT: {dataset_name}")
    print(f"{'=' * 40}")

    print("Total rows:", len(data))
    print("Total columns:", len(data.columns))
    print("Full duplicate rows:", data.duplicated().sum())

    print("\nMissing values by column:")
    missing_values = data.isnull().sum()
    print(missing_values[missing_values > 0])

    print("\nDuplicate rows by column key:")
    for column in data.columns:
        if column.endswith("_id"):
            print(
                f"{column}:",
                data[column].duplicated().sum()
            )


# Check both datasets
check_data_quality(payment_methods, "Payment Methods")
check_data_quality(transactions, "Transactions")
```
